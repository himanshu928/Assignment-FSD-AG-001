# Question 1.

Explain the basic structure of an HTML page. Why is it important to follow
this structure?

# Answer
-  HTML (Hyper Text Markup Language) is the standard language used to create web pages.
It follows the basic structure which helps the browser understand how to display the content properly in web pages. It contains some basic and important elements that decides how web pages look like.
By following this structure, developers can build well organized, efficient, and browser-friendly websites.


```python
# Answer 1.

# Example of Basic HTML Structure :
<!Doctype html>
<html>
<head>
       <title>Web page</title>
</head>
<body>
       <h1>Welcome to My website</h1>
       <p>This is my first HTML Page.</p>
</body>
</html>
```

# Question 3.
Look at the code below. Can you find and fix the mistake?

# Answer
- In the given code, there is a mistake in the HTML comment.
The comment is not closed properly, so the browser will treat everything after it as part of the comment and the heading will not be displayed.

# Question 4.

You’re creating a form that asks the user to enter their name, select their country from a list, and click a button to submit. Write the HTML code for this.

# Answer

- To create a form where a user can enter their name, select their country from a list, and submit the form, we use HTML form elements such as < form>, < input>, < select>, and < button>.

```python
# Answer 4.

<form>
  <label for="name">Enter your name:</label><br>
  <input type="text" id="name"><br><br>

  <label for="country">Select your country:</label><br>
  <select id="country">
    <option value="India">India</option>
    <option value="USA">USA</option>
    <option value="UK">UK</option>
    <option value="Russia">Russia</option>
    <option value="Israil">Israil</option>
  </select><br><br>

  <button type="submit">Submit</button>
</form>
```
